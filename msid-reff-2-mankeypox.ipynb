```python
# ============================================================
# FULL KAGGLE NOTEBOOK (Monkeypox multi-class)
# Ensemble CNN backbones + lightweight ViT block
# + Confusion Matrix + ROC
# + Bootstrap 95% Confidence Intervals for performance metrics
# ============================================================

import os
import random
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models, Input
from tensorflow.keras.applications import ResNet50, EfficientNetB0, DenseNet121, Xception
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet_pre
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.utils import to_categorical

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, auc,
    classification_report, cohen_kappa_score,
    precision_recall_fscore_support, accuracy_score,
    roc_auc_score
)

# -----------------------------
# CONFIG
# -----------------------------
SEED = 42
IMAGE_SIZE = (224, 224)
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 100
SAMPLES_PER_CLASS = 1000
DATA_DIR = "/kaggle/input/msid-ref-2/MSID/Monkeypox Skin Image Dataset"  # update if needed

# -----------------------------
# REPRODUCIBILITY
# -----------------------------
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# -----------------------------
# LOAD & BALANCE DATA IN MEMORY
# -----------------------------
print("📦 Loading and sampling images per class...")

class_names = sorted([d for d in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, d))])
print("Classes:", class_names)

all_images, all_labels_idx = [], []

for idx, class_name in enumerate(class_names):
    class_dir = os.path.join(DATA_DIR, class_name)
    image_files = [f for f in os.listdir(class_dir) if f.lower().endswith((".jpg", ".jpeg", ".png"))]

    if len(image_files) == 0:
        print(f"⚠️ No images found in: {class_dir}")
        continue

    selected = random.sample(image_files, min(SAMPLES_PER_CLASS, len(image_files)))

    for file in selected:
        img = load_img(os.path.join(class_dir, file), target_size=IMAGE_SIZE)
        img_array = img_to_array(img)
        img_array = resnet_pre(img_array)  # using ResNet preprocess for all (OK, but you can also match each backbone)
        all_images.append(img_array)
        all_labels_idx.append(idx)

all_images = np.array(all_images, dtype=np.float32)
all_labels_idx = np.array(all_labels_idx, dtype=np.int32)
all_labels_oh = to_categorical(all_labels_idx, num_classes=NUM_CLASSES)

print("Total images:", len(all_images))

# -----------------------------
# SPLIT (stratify must use class indices, NOT one-hot)
# -----------------------------
X_train, X_val, y_train, y_val, y_train_idx, y_val_idx = train_test_split(
    all_images, all_labels_oh, all_labels_idx,
    test_size=0.2,
    stratify=all_labels_idx,
    random_state=SEED
)

print("Train:", X_train.shape, y_train.shape)
print("Val  :", X_val.shape, y_val.shape)

# -----------------------------
# ViT BLOCK (small)
# -----------------------------
def create_vit_block(input_shape, projection_dim=64, num_heads=4, transformer_layers=4, patch_size=16):
    inputs = Input(shape=input_shape)

    # patch embedding
    num_patches = (input_shape[0] // patch_size) * (input_shape[1] // patch_size)
    x = layers.Conv2D(filters=projection_dim, kernel_size=patch_size, strides=patch_size, padding="valid")(inputs)
    x = layers.Reshape((num_patches, projection_dim))(x)

    # position embedding
    positions = tf.range(start=0, limit=num_patches, delta=1)
    pos_embed = layers.Embedding(input_dim=num_patches, output_dim=projection_dim)(positions)
    x = x + pos_embed

    # transformer encoder
    for _ in range(transformer_layers):
        x1 = layers.LayerNormalization(epsilon=1e-6)(x)
        attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=projection_dim)(x1, x1)
        x2 = layers.Add()([x, attn])

        x3 = layers.LayerNormalization(epsilon=1e-6)(x2)
        ffn = tf.keras.Sequential([
            layers.Dense(4 * projection_dim, activation="gelu"),
            layers.Dropout(0.1),
            layers.Dense(projection_dim),
        ])(x3)
        x = layers.Add()([x2, ffn])

    x = layers.LayerNormalization(epsilon=1e-6)(x)
    x = layers.GlobalAveragePooling1D()(x)
    return models.Model(inputs, x, name="ViT_Block")

# -----------------------------
# FEATURE EXTRACTION BACKBONES
# -----------------------------
input_layer = Input(shape=(224, 224, 3))

resnet = ResNet50(include_top=False, weights="imagenet", input_shape=(224, 224, 3))
efficientnet = EfficientNetB0(include_top=False, weights="imagenet", input_shape=(224, 224, 3))
densenet = DenseNet121(include_top=False, weights="imagenet", input_shape=(224, 224, 3))
xception = Xception(include_top=False, weights="imagenet", input_shape=(224, 224, 3))

for base in [resnet, efficientnet, densenet, xception]:
    base.trainable = False

def extract_features(base_model, x):
    x = base_model(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    return x

resnet_feat = extract_features(resnet, input_layer)
efficientnet_feat = extract_features(efficientnet, input_layer)
densenet_feat = extract_features(densenet, input_layer)
xception_feat = extract_features(xception, input_layer)

vit_feat = create_vit_block((224, 224, 3))(input_layer)

# -----------------------------
# FUSION + CLASSIFIER
# -----------------------------
combined = layers.Concatenate()([resnet_feat, efficientnet_feat, densenet_feat, xception_feat, vit_feat])

x = layers.Dense(512, activation="relu")(combined)
x = layers.Dropout(0.5)(x)
output = layers.Dense(NUM_CLASSES, activation="softmax")(x)

model = models.Model(inputs=input_layer, outputs=output)
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# -----------------------------
# TRAIN
# -----------------------------
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=12, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6)
]

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=callbacks,
    verbose=1
)

# -----------------------------
# PLOT ACCURACY & LOSS
# -----------------------------
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Train")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("Accuracy")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Train")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# -----------------------------
# EVALUATION (VAL AS TEST)
# -----------------------------
test_loss, test_acc = model.evaluate(X_val, y_val, batch_size=32, verbose=0)
print(f"✅ Test Accuracy (Val): {test_acc*100:.2f}%")

y_pred = model.predict(X_val, batch_size=32, verbose=0)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_val, axis=1)

# -----------------------------
# CONFUSION MATRIX
# -----------------------------
cm = confusion_matrix(y_true_classes, y_pred_classes)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
disp.plot(xticks_rotation=45, cmap="Blues")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()

# -----------------------------
# ROC CURVE (one-vs-rest per class)
# -----------------------------
plt.figure(figsize=(10, 7))
for i in range(NUM_CLASSES):
    fpr, tpr, _ = roc_curve(y_val[:, i], y_pred[:, i])
    auc_score = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f"{class_names[i]} (AUC={auc_score:.2f})")

plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC Curve (OvR)")
plt.legend()
plt.grid(True)
plt.show()

# -----------------------------
# CLASSIFICATION REPORT + KAPPA + SENSITIVITY
# -----------------------------
report = classification_report(y_true_classes, y_pred_classes, target_names=class_names, digits=4)
print("📊 Classification Report:\n", report)

kappa = cohen_kappa_score(y_true_classes, y_pred_classes)
print(f"✅ Cohen's Kappa Score: {kappa:.4f}")

sensitivity_per_class = precision_recall_fscore_support(y_true_classes, y_pred_classes, average=None)[1]
for idx, cls in enumerate(class_names):
    print(f"🔹 Sensitivity (Recall) for {cls}: {sensitivity_per_class[idx]:.4f}")

precision_micro, recall_micro, f1_micro, _ = precision_recall_fscore_support(
    y_true_classes, y_pred_classes, average="micro"
)
precision_macro, recall_macro, f1_macro, _ = precision_recall_fscore_support(
    y_true_classes, y_pred_classes, average="macro"
)
precision_weighted, recall_weighted, f1_weighted, _ = precision_recall_fscore_support(
    y_true_classes, y_pred_classes, average="weighted"
)

print("\n📌 Averages:")
print(f"Micro Average    -> Precision: {precision_micro:.4f}, Recall: {recall_micro:.4f}, F1-score: {f1_micro:.4f}")
print(f"Macro Average    -> Precision: {precision_macro:.4f}, Recall: {recall_macro:.4f}, F1-score: {f1_macro:.4f}")
print(f"Weighted Average -> Precision: {precision_weighted:.4f}, Recall: {recall_weighted:.4f}, F1-score: {f1_weighted:.4f}")

# ============================================================
# BOOTSTRAP 95% CONFIDENCE INTERVALS
# ============================================================
def bootstrap_ci(y_true_cls, y_prob, n_boot=2000, ci=95, seed=42):
    """
    y_true_cls: (N,) int labels
    y_prob    : (N, C) predicted probabilities
    Returns dict of metric -> (mean, low, high)
    """
    rng = np.random.default_rng(seed)
    n = len(y_true_cls)

    acc_list = []
    prec_macro_list = []
    rec_macro_list = []
    f1_macro_list = []
    kappa_list = []
    auc_ovr_macro_list = []

    # Convert true labels to one-hot for AUC
    y_true_oh = to_categorical(y_true_cls, num_classes=y_prob.shape[1])

    for _ in range(n_boot):
        idx = rng.integers(0, n, size=n)  # sample with replacement
        yt = y_true_cls[idx]
        yp = y_prob[idx]
        yp_cls = np.argmax(yp, axis=1)

        acc_list.append(accuracy_score(yt, yp_cls))

        p, r, f1, _ = precision_recall_fscore_support(yt, yp_cls, average="macro", zero_division=0)
        prec_macro_list.append(p)
        rec_macro_list.append(r)
        f1_macro_list.append(f1)

        kappa_list.append(cohen_kappa_score(yt, yp_cls))

        # AUC OvR macro (needs at least one sample per class in this bootstrap draw)
        # Handle rare failure safely:
        try:
            yt_oh = to_categorical(yt, num_classes=yp.shape[1])
            auc_ovr_macro_list.append(
                roc_auc_score(yt_oh, yp, multi_class="ovr", average="macro")
            )
        except Exception:
            # skip this draw if AUC cannot be computed
            pass

    def ci_bounds(values):
        values = np.array(values, dtype=np.float64)
        mean = float(np.mean(values))
        alpha = (100 - ci) / 2
        low = float(np.percentile(values, alpha))
        high = float(np.percentile(values, 100 - alpha))
        return mean, low, high

    out = {
        "Accuracy": ci_bounds(acc_list),
        "Precision (macro)": ci_bounds(prec_macro_list),
        "Recall (macro)": ci_bounds(rec_macro_list),
        "F1 (macro)": ci_bounds(f1_macro_list),
        "Cohen's Kappa": ci_bounds(kappa_list),
    }
    if len(auc_ovr_macro_list) > 50:
        out["AUC (ovr-macro)"] = ci_bounds(auc_ovr_macro_list)
    else:
        out["AUC (ovr-macro)"] = ("NA", "NA", "NA")

    return out

ci_results = bootstrap_ci(y_true_classes, y_pred, n_boot=2000, ci=95, seed=SEED)

print("\n================ BOOTSTRAP 95% CI (Val set) ================")
for k, v in ci_results.items():
    if v[0] == "NA":
        print(f"{k}: NA (not enough valid bootstrap samples)")
    else:
        mean, low, high = v
        print(f"{k}: {mean:.4f} (95% CI: {low:.4f} – {high:.4f})")

# Also print the single-shot AUC (ovr-macro) on full val (not bootstrap)
try:
    auc_full = roc_auc_score(y_val, y_pred, multi_class="ovr", average="macro")
    print(f"\nAUC (ovr-macro) on full Val: {auc_full:.4f}")
except Exception as e:
    print("\nAUC (ovr-macro) on full Val: could not compute ->", e)
```

```python
# ============================================================
# FULL KAGGLE NOTEBOOK (Monkeypox multi-class)
# Ensemble CNN backbones + lightweight ViT block
# + Confusion Matrix + ROC
# + Bootstrap 95% Confidence Intervals for performance metrics
# ============================================================

import os
import random
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models, Input
from tensorflow.keras.applications import ResNet50, EfficientNetB0, DenseNet121, Xception
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet_pre
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.utils import to_categorical

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, auc,
    classification_report, cohen_kappa_score,
    precision_recall_fscore_support, accuracy_score,
    roc_auc_score
)

# -----------------------------
# CONFIG
# -----------------------------
SEED = 42
IMAGE_SIZE = (224, 224)
NUM_CLASSES = 4
BATCH_SIZE = 32
EPOCHS = 100
SAMPLES_PER_CLASS = 1000
DATA_DIR = "/kaggle/input/msid-ref-2/MSID/Monkeypox Skin Image Dataset"  # update if needed

# -----------------------------
# REPRODUCIBILITY
# -----------------------------
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# -----------------------------
# LOAD & BALANCE DATA IN MEMORY
# -----------------------------
print("📦 Loading and sampling images per class...")

class_names = sorted([d for d in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, d))])
print("Classes:", class_names)

all_images, all_labels_idx = [], []

for idx, class_name in enumerate(class_names):
    class_dir = os.path.join(DATA_DIR, class_name)
    image_files = [f for f in os.listdir(class_dir) if f.lower().endswith((".jpg", ".jpeg", ".png"))]

    if len(image_files) == 0:
        print(f"⚠️ No images found in: {class_dir}")
        continue

    selected = random.sample(image_files, min(SAMPLES_PER_CLASS, len(image_files)))

    for file in selected:
        img = load_img(os.path.join(class_dir, file), target_size=IMAGE_SIZE)
        img_array = img_to_array(img)
        img_array = resnet_pre(img_array)  # using ResNet preprocess for all (OK, but you can also match each backbone)
        all_images.append(img_array)
        all_labels_idx.append(idx)

all_images = np.array(all_images, dtype=np.float32)
all_labels_idx = np.array(all_labels_idx, dtype=np.int32)
all_labels_oh = to_categorical(all_labels_idx, num_classes=NUM_CLASSES)

print("Total images:", len(all_images))

# -----------------------------
# SPLIT (stratify must use class indices, NOT one-hot)
# -----------------------------
X_train, X_val, y_train, y_val, y_train_idx, y_val_idx = train_test_split(
    all_images, all_labels_oh, all_labels_idx,
    test_size=0.2,
    stratify=all_labels_idx,
    random_state=SEED
)

print("Train:", X_train.shape, y_train.shape)
print("Val  :", X_val.shape, y_val.shape)

# -----------------------------
# ViT BLOCK (small)
# -----------------------------
def create_vit_block(input_shape, projection_dim=64, num_heads=4, transformer_layers=4, patch_size=16):
    inputs = Input(shape=input_shape)

    # patch embedding
    num_patches = (input_shape[0] // patch_size) * (input_shape[1] // patch_size)
    x = layers.Conv2D(filters=projection_dim, kernel_size=patch_size, strides=patch_size, padding="valid")(inputs)
    x = layers.Reshape((num_patches, projection_dim))(x)

    # position embedding
    positions = tf.range(start=0, limit=num_patches, delta=1)
    pos_embed = layers.Embedding(input_dim=num_patches, output_dim=projection_dim)(positions)
    x = x + pos_embed

    # transformer encoder
    for _ in range(transformer_layers):
        x1 = layers.LayerNormalization(epsilon=1e-6)(x)
        attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=projection_dim)(x1, x1)
        x2 = layers.Add()([x, attn])

        x3 = layers.LayerNormalization(epsilon=1e-6)(x2)
        ffn = tf.keras.Sequential([
            layers.Dense(4 * projection_dim, activation="gelu"),
            layers.Dropout(0.1),
            layers.Dense(projection_dim),
        ])(x3)
        x = layers.Add()([x2, ffn])

    x = layers.LayerNormalization(epsilon=1e-6)(x)
    x = layers.GlobalAveragePooling1D()(x)
    return models.Model(inputs, x, name="ViT_Block")

# -----------------------------
# FEATURE EXTRACTION BACKBONES
# -----------------------------
input_layer = Input(shape=(224, 224, 3))

resnet = ResNet50(include_top=False, weights="imagenet", input_shape=(224, 224, 3))
efficientnet = EfficientNetB0(include_top=False, weights="imagenet", input_shape=(224, 224, 3))
densenet = DenseNet121(include_top=False, weights="imagenet", input_shape=(224, 224, 3))
xception = Xception(include_top=False, weights="imagenet", input_shape=(224, 224, 3))

for base in [resnet, efficientnet, densenet, xception]:
    base.trainable = False

def extract_features(base_model, x):
    x = base_model(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    return x

resnet_feat = extract_features(resnet, input_layer)
efficientnet_feat = extract_features(efficientnet, input_layer)
densenet_feat = extract_features(densenet, input_layer)
xception_feat = extract_features(xception, input_layer)

vit_feat = create_vit_block((224, 224, 3))(input_layer)

# -----------------------------
# FUSION + CLASSIFIER
# -----------------------------
combined = layers.Concatenate()([resnet_feat, efficientnet_feat, densenet_feat, xception_feat, vit_feat])

x = layers.Dense(512, activation="relu")(combined)
x = layers.Dropout(0.5)(x)
output = layers.Dense(NUM_CLASSES, activation="softmax")(x)

model = models.Model(inputs=input_layer, outputs=output)
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# -----------------------------
# TRAIN
# -----------------------------
callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=12, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6)
]

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=callbacks,
    verbose=1
)

# -----------------------------
# PLOT ACCURACY & LOSS
# -----------------------------
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Train")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("Accuracy")
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Train")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# -----------------------------
# EVALUATION (VAL AS TEST)
# -----------------------------
test_loss, test_acc = model.evaluate(X_val, y_val, batch_size=32, verbose=0)
print(f"✅ Test Accuracy (Val): {test_acc*100:.2f}%")

y_pred = model.predict(X_val, batch_size=32, verbose=0)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_val, axis=1)

# -----------------------------
# CONFUSION MATRIX
# -----------------------------
cm = confusion_matrix(y_true_classes, y_pred_classes)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
disp.plot(xticks_rotation=45, cmap="Blues")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()

# -----------------------------
# ROC CURVE (one-vs-rest per class)
# -----------------------------
plt.figure(figsize=(10, 7))
for i in range(NUM_CLASSES):
    fpr, tpr, _ = roc_curve(y_val[:, i], y_pred[:, i])
    auc_score = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f"{class_names[i]} (AUC={auc_score:.2f})")

plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC Curve (OvR)")
plt.legend()
plt.grid(True)
plt.show()

# -----------------------------
# CLASSIFICATION REPORT + KAPPA + SENSITIVITY
# -----------------------------
report = classification_report(y_true_classes, y_pred_classes, target_names=class_names, digits=4)
print("📊 Classification Report:\n", report)

kappa = cohen_kappa_score(y_true_classes, y_pred_classes)
print(f"✅ Cohen's Kappa Score: {kappa:.4f}")

sensitivity_per_class = precision_recall_fscore_support(y_true_classes, y_pred_classes, average=None)[1]
for idx, cls in enumerate(class_names):
    print(f"🔹 Sensitivity (Recall) for {cls}: {sensitivity_per_class[idx]:.4f}")

precision_micro, recall_micro, f1_micro, _ = precision_recall_fscore_support(
    y_true_classes, y_pred_classes, average="micro"
)
precision_macro, recall_macro, f1_macro, _ = precision_recall_fscore_support(
    y_true_classes, y_pred_classes, average="macro"
)
precision_weighted, recall_weighted, f1_weighted, _ = precision_recall_fscore_support(
    y_true_classes, y_pred_classes, average="weighted"
)

print("\n📌 Averages:")
print(f"Micro Average    -> Precision: {precision_micro:.4f}, Recall: {recall_micro:.4f}, F1-score: {f1_micro:.4f}")
print(f"Macro Average    -> Precision: {precision_macro:.4f}, Recall: {recall_macro:.4f}, F1-score: {f1_macro:.4f}")
print(f"Weighted Average -> Precision: {precision_weighted:.4f}, Recall: {recall_weighted:.4f}, F1-score: {f1_weighted:.4f}")

# ============================================================
# BOOTSTRAP 95% CONFIDENCE INTERVALS
# ============================================================
def bootstrap_ci(y_true_cls, y_prob, n_boot=2000, ci=95, seed=42):
    """
    y_true_cls: (N,) int labels
    y_prob    : (N, C) predicted probabilities
    Returns dict of metric -> (mean, low, high)
    """
    rng = np.random.default_rng(seed)
    n = len(y_true_cls)

    acc_list = []
    prec_macro_list = []
    rec_macro_list = []
    f1_macro_list = []
    kappa_list = []
    auc_ovr_macro_list = []

    # Convert true labels to one-hot for AUC
    y_true_oh = to_categorical(y_true_cls, num_classes=y_prob.shape[1])

    for _ in range(n_boot):
        idx = rng.integers(0, n, size=n)  # sample with replacement
        yt = y_true_cls[idx]
        yp = y_prob[idx]
        yp_cls = np.argmax(yp, axis=1)

        acc_list.append(accuracy_score(yt, yp_cls))

        p, r, f1, _ = precision_recall_fscore_support(yt, yp_cls, average="macro", zero_division=0)
        prec_macro_list.append(p)
        rec_macro_list.append(r)
        f1_macro_list.append(f1)

        kappa_list.append(cohen_kappa_score(yt, yp_cls))

        # AUC OvR macro (needs at least one sample per class in this bootstrap draw)
        # Handle rare failure safely:
        try:
            yt_oh = to_categorical(yt, num_classes=yp.shape[1])
            auc_ovr_macro_list.append(
                roc_auc_score(yt_oh, yp, multi_class="ovr", average="macro")
            )
        except Exception:
            # skip this draw if AUC cannot be computed
            pass

    def ci_bounds(values):
        values = np.array(values, dtype=np.float64)
        mean = float(np.mean(values))
        alpha = (100 - ci) / 2
        low = float(np.percentile(values, alpha))
        high = float(np.percentile(values, 100 - alpha))
        return mean, low, high

    out = {
        "Accuracy": ci_bounds(acc_list),
        "Precision (macro)": ci_bounds(prec_macro_list),
        "Recall (macro)": ci_bounds(rec_macro_list),
        "F1 (macro)": ci_bounds(f1_macro_list),
        "Cohen's Kappa": ci_bounds(kappa_list),
    }
    if len(auc_ovr_macro_list) > 50:
        out["AUC (ovr-macro)"] = ci_bounds(auc_ovr_macro_list)
    else:
        out["AUC (ovr-macro)"] = ("NA", "NA", "NA")

    return out

ci_results = bootstrap_ci(y_true_classes, y_pred, n_boot=2000, ci=95, seed=SEED)

print("\n================ BOOTSTRAP 95% CI (Val set) ================")
for k, v in ci_results.items():
    if v[0] == "NA":
        print(f"{k}: NA (not enough valid bootstrap samples)")
    else:
        mean, low, high = v
        print(f"{k}: {mean:.4f} (95% CI: {low:.4f} – {high:.4f})")

# Also print the single-shot AUC (ovr-macro) on full val (not bootstrap)
try:
    auc_full = roc_auc_score(y_val, y_pred, multi_class="ovr", average="macro")
    print(f"\nAUC (ovr-macro) on full Val: {auc_full:.4f}")
except Exception as e:
    print("\nAUC (ovr-macro) on full Val: could not compute ->", e)
```

```python
# ============================================================
# ✅ FULL KAGGLE NOTEBOOK (MSID / Monkeypox 4-class)
# ✅ Ensemble CNN backbones + lightweight ViT block
# ✅ 5-FOLD STRATIFIED CV
# ✅ Two-stage training (warm-up + fine-tune)
# ✅ Confusion Matrix + ROC + Report + Kappa + Sensitivity
# ✅ Bootstrap 95% CI per fold
#
# FIXED ISSUES:
# 1) Name collision error: solved by NOT using input_tensor=
# 2) HTTP 403 on EfficientNet weights: handled with safe loader fallback
# ============================================================

import os, random
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models, Input
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.utils import to_categorical

from tensorflow.keras.applications import ResNet50, DenseNet121, Xception, EfficientNetB0

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, auc,
    classification_report, cohen_kappa_score,
    precision_recall_fscore_support, accuracy_score,
    roc_auc_score
)
from sklearn.utils.class_weight import compute_class_weight

# -----------------------------
# CONFIG
# -----------------------------
SEED = 42
IMAGE_SIZE = (224, 224)
NUM_CLASSES = 4
BATCH_SIZE = 32

# Stage A warmup
EPOCHS_WARMUP = 10
LR_WARMUP = 1e-4

# Stage B finetune
EPOCHS_FINETUNE = 25
LR_FINETUNE = 1e-5
UNFREEZE_LAST_N = 30

# Data
SAMPLES_PER_CLASS = 1000  # set None to use all
DATA_DIR = "/kaggle/input/msid-ref-2/MSID/Monkeypox Skin Image Dataset"

# CV + CI
N_SPLITS = 5
BOOT_N = 2000
CI_LEVEL = 95

# Backbones toggles (to avoid 403 / memory)
USE_EFFICIENTNET = False   # ✅ safest: set False to avoid 403
USE_XCEPTION     = True

# -----------------------------
# REPRODUCIBILITY
# -----------------------------
os.environ["PYTHONHASHSEED"] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# -----------------------------
# LOAD DATA (NO AUGMENTATION HERE)
# -----------------------------
print("📦 Loading images...")

class_names = sorted([d for d in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, d))])
print("Classes:", class_names)
assert len(class_names) == NUM_CLASSES, f"Found {len(class_names)} classes, expected {NUM_CLASSES}"

X_list, y_idx_list = [], []

for cls_idx, cls in enumerate(class_names):
    cls_dir = os.path.join(DATA_DIR, cls)
    files = [f for f in os.listdir(cls_dir) if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    if len(files) == 0:
        raise ValueError(f"No images found in {cls_dir}")

    if SAMPLES_PER_CLASS is not None:
        files = random.sample(files, min(SAMPLES_PER_CLASS, len(files)))

    for f in files:
        img = load_img(os.path.join(cls_dir, f), target_size=IMAGE_SIZE)
        arr = img_to_array(img).astype(np.float32) / 255.0
        X_list.append(arr)
        y_idx_list.append(cls_idx)

X = np.stack(X_list, axis=0)
y_idx = np.array(y_idx_list, dtype=np.int32)

print("Total images:", len(X))
print("Class counts:", {class_names[i]: int(np.sum(y_idx == i)) for i in range(NUM_CLASSES)})

# -----------------------------
# TRAIN-ONLY AUGMENTATION
# -----------------------------
train_aug = tf.keras.Sequential(
    [
        layers.RandomFlip("horizontal"),
        layers.RandomRotation(0.08),
        layers.RandomZoom(0.10),
        layers.RandomContrast(0.10),
    ],
    name="train_augmentation"
)

# -----------------------------
# LIGHTWEIGHT ViT BLOCK
# -----------------------------
def create_vit_block(input_shape, projection_dim=64, num_heads=4, transformer_layers=3, patch_size=16):
    inputs = Input(shape=input_shape)

    num_patches = (input_shape[0] // patch_size) * (input_shape[1] // patch_size)
    x = layers.Conv2D(projection_dim, kernel_size=patch_size, strides=patch_size, padding="valid")(inputs)
    x = layers.Reshape((num_patches, projection_dim))(x)

    positions = tf.range(start=0, limit=num_patches, delta=1)
    pos_embed = layers.Embedding(input_dim=num_patches, output_dim=projection_dim)(positions)
    x = x + pos_embed

    for _ in range(transformer_layers):
        x1 = layers.LayerNormalization(epsilon=1e-6)(x)
        attn = layers.MultiHeadAttention(num_heads=num_heads, key_dim=projection_dim)(x1, x1)
        x2 = layers.Add()([x, attn])

        x3 = layers.LayerNormalization(epsilon=1e-6)(x2)
        ffn = tf.keras.Sequential(
            [
                layers.Dense(4 * projection_dim, activation="gelu"),
                layers.Dropout(0.1),
                layers.Dense(projection_dim),
            ]
        )(x3)
        x = layers.Add()([x2, ffn])

    x = layers.LayerNormalization(epsilon=1e-6)(x)
    x = layers.GlobalAveragePooling1D()(x)
    return models.Model(inputs, x, name="ViT_Block")

# -----------------------------
# SAFE BACKBONE LOADER (handles HTTP 403)
# -----------------------------
def safe_backbone(constructor, name, input_shape=(224,224,3), prefer_imagenet=True):
    """
    Try weights='imagenet'. If download fails (403), fall back to weights=None.
    """
    if prefer_imagenet:
        try:
            return constructor(include_top=False, weights="imagenet", input_shape=input_shape, name=name)
        except Exception as e:
            print(f"⚠️ Could not load ImageNet weights for {name}: {e}")
            print(f"➡️ Falling back to weights=None for {name}")
    return constructor(include_top=False, weights=None, input_shape=input_shape, name=name)

# -----------------------------
# BUILD MODEL (PER FOLD)
# -----------------------------
def build_model(num_classes=NUM_CLASSES, label_smoothing=0.05):
    inp = Input(shape=(224, 224, 3), name="input")
    x_aug = train_aug(inp)

    # ✅ Build without input_tensor to avoid name collisions
    resnet = safe_backbone(ResNet50,   "ResNet50_backbone",   prefer_imagenet=True)
    dense  = safe_backbone(DenseNet121,"DenseNet121_backbone",prefer_imagenet=True)

    backbones = [resnet, dense]

    if USE_EFFICIENTNET:
        eff = safe_backbone(EfficientNetB0, "EffNetB0_backbone", prefer_imagenet=True)
        backbones.append(eff)

    if USE_XCEPTION:
        xcep = safe_backbone(Xception, "Xception_backbone", prefer_imagenet=True)
        backbones.append(xcep)

    # Freeze warmup
    for b in backbones:
        b.trainable = False

    def feat(backbone, x):
        y = backbone(x, training=False)
        y = layers.GlobalAveragePooling2D()(y)
        return y

    feats = [feat(b, x_aug) for b in backbones]

    vit = create_vit_block((224, 224, 3), projection_dim=64, num_heads=4, transformer_layers=3, patch_size=16)
    feats.append(vit(x_aug))

    fused = layers.Concatenate(name="fusion")(feats)

    z = layers.BatchNormalization()(fused)
    z = layers.Dense(1024, activation="relu")(z)
    z = layers.Dropout(0.4)(z)
    z = layers.Dense(256, activation="relu")(z)
    z = layers.Dropout(0.3)(z)
    out = layers.Dense(num_classes, activation="softmax")(z)

    model = models.Model(inp, out, name="EnsembleCNN_ViT")

    loss_fn = tf.keras.losses.CategoricalCrossentropy(label_smoothing=label_smoothing)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(LR_WARMUP),
        loss=loss_fn,
        metrics=["accuracy"]
    )
    return model, tuple(backbones)

def unfreeze_last_n(backbones, n=30):
    for b in backbones:
        b.trainable = True
        for layer in b.layers[:-n]:
            layer.trainable = False
        # keep BN frozen for stability
        for layer in b.layers:
            if isinstance(layer, layers.BatchNormalization):
                layer.trainable = False

# -----------------------------
# BOOTSTRAP CI
# -----------------------------
def bootstrap_ci(y_true_cls, y_prob, n_boot=2000, ci=95, seed=42):
    rng = np.random.default_rng(seed)
    n = len(y_true_cls)

    acc_list, prec_list, rec_list, f1_list, kappa_list, auc_list = [], [], [], [], [], []

    for _ in range(n_boot):
        idx = rng.integers(0, n, size=n)
        yt = y_true_cls[idx]
        yp = y_prob[idx]
        yp_cls = np.argmax(yp, axis=1)

        acc_list.append(accuracy_score(yt, yp_cls))
        p, r, f1, _ = precision_recall_fscore_support(yt, yp_cls, average="macro", zero_division=0)
        prec_list.append(p); rec_list.append(r); f1_list.append(f1)
        kappa_list.append(cohen_kappa_score(yt, yp_cls))

        try:
            yt_oh = to_categorical(yt, num_classes=yp.shape[1])
            auc_list.append(roc_auc_score(yt_oh, yp, multi_class="ovr", average="macro"))
        except Exception:
            pass

    def bounds(vals):
        vals = np.array(vals, dtype=np.float64)
        alpha = (100 - ci) / 2
        return float(np.mean(vals)), float(np.percentile(vals, alpha)), float(np.percentile(vals, 100 - alpha))

    out = {
        "Accuracy": bounds(acc_list),
        "Precision (macro)": bounds(prec_list),
        "Recall (macro)": bounds(rec_list),
        "F1 (macro)": bounds(f1_list),
        "Cohen's Kappa": bounds(kappa_list),
    }
    out["AUC (ovr-macro)"] = bounds(auc_list) if len(auc_list) > 50 else ("NA", "NA", "NA")
    return out

# -----------------------------
# 5-FOLD STRATIFIED CV
# -----------------------------
skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=SEED)
fold_metrics = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y_idx), start=1):
    print(f"\n\n====================== FOLD {fold}/{N_SPLITS} ======================")

    X_train, X_val = X[train_idx], X[val_idx]
    y_train_idx, y_val_idx = y_idx[train_idx], y_idx[val_idx]
    print("Val class counts:", {class_names[i]: int(np.sum(y_val_idx == i)) for i in range(NUM_CLASSES)})

    y_train = to_categorical(y_train_idx, num_classes=NUM_CLASSES)
    y_val = to_categorical(y_val_idx, num_classes=NUM_CLASSES)

    cw = compute_class_weight(class_weight="balanced", classes=np.arange(NUM_CLASSES), y=y_train_idx)
    class_weight = {i: float(cw[i]) for i in range(NUM_CLASSES)}
    print("Class weights:", class_weight)

    tf.keras.backend.clear_session()
    model, backbones = build_model()

    callbacks = [
        tf.keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=8, restore_best_weights=True),
        tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-6),
    ]

    # Stage A warmup
    hist1 = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS_WARMUP,
        batch_size=BATCH_SIZE,
        callbacks=callbacks,
        class_weight=class_weight,
        verbose=1
    )

    # Stage B fine-tune
    unfreeze_last_n(backbones, n=UNFREEZE_LAST_N)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(LR_FINETUNE),
        loss=tf.keras.losses.CategoricalCrossentropy(label_smoothing=0.05),
        metrics=["accuracy"]
    )
    hist2 = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=EPOCHS_FINETUNE,
        batch_size=BATCH_SIZE,
        callbacks=callbacks,
        class_weight=class_weight,
        verbose=1
    )

    # Evaluate
    loss, acc = model.evaluate(X_val, y_val, batch_size=32, verbose=0)
    print(f"✅ Fold {fold} Accuracy: {acc*100:.2f}%")

    y_prob = model.predict(X_val, batch_size=32, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)
    y_true = y_val_idx

    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
    disp.plot(xticks_rotation=45, cmap="Blues")
    plt.title(f"Fold {fold} Confusion Matrix")
    plt.tight_layout()
    plt.show()

    # ROC
    plt.figure(figsize=(10,7))
    for i in range(NUM_CLASSES):
        fpr, tpr, _ = roc_curve(y_val[:, i], y_prob[:, i])
        auc_score = auc(fpr, tpr)
        plt.plot(fpr, tpr, label=f"{class_names[i]} (AUC={auc_score:.2f})")
    plt.plot([0,1],[0,1],"k--")
    plt.xlabel("FPR"); plt.ylabel("TPR")
    plt.title(f"Fold {fold} ROC (OvR)")
    plt.legend(); plt.grid(True)
    plt.show()

    # Report + Kappa + Sensitivity
    print("📊 Classification Report:\n",
          classification_report(y_true, y_pred, target_names=class_names, digits=4))

    kappa = cohen_kappa_score(y_true, y_pred)
    print(f"✅ Fold {fold} Cohen's Kappa: {kappa:.4f}")

    sens = precision_recall_fscore_support(y_true, y_pred, average=None, zero_division=0)[1]
    for i, cls in enumerate(class_names):
        print(f"🔹 Sensitivity (Recall) for {cls}: {sens[i]:.4f}")

    # Bootstrap CI
    ci_results = bootstrap_ci(y_true, y_prob, n_boot=BOOT_N, ci=CI_LEVEL, seed=SEED + fold)
    print(f"\n================ BOOTSTRAP {CI_LEVEL}% CI (Fold {fold}) ================")
    for k, v in ci_results.items():
        if v[0] == "NA":
            print(f"{k}: NA")
        else:
            m, lo, hi = v
            print(f"{k}: {m:.4f} ({CI_LEVEL}% CI: {lo:.4f} – {hi:.4f})")

    # Store fold metrics
    prec_macro, rec_macro, f1_macro, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )
    try:
        auc_macro = roc_auc_score(y_val, y_prob, multi_class="ovr", average="macro")
    except Exception:
        auc_macro = np.nan

    fold_metrics.append({
        "fold": fold,
        "acc": float(accuracy_score(y_true, y_pred)),
        "prec_macro": float(prec_macro),
        "rec_macro": float(rec_macro),
        "f1_macro": float(f1_macro),
        "kappa": float(kappa),
        "auc_macro": float(auc_macro) if np.isfinite(auc_macro) else np.nan
    })

# -----------------------------
# FINAL SUMMARY
# -----------------------------
import pandas as pd
dfm = pd.DataFrame(fold_metrics)

print("\n\n==================== 5-FOLD CV SUMMARY ====================")
for col in ["acc", "prec_macro", "rec_macro", "f1_macro", "kappa", "auc_macro"]:
    vals = dfm[col].values.astype(float)
    print(f"{col:12s}: {np.nanmean(vals):.4f} ± {np.nanstd(vals):.4f}")

print("\nPer-fold table:")
print(dfm)

print("\n✅ NOTE:")
print("If you want higher accuracy, try: UNFREEZE_LAST_N=60 and EPOCHS_FINETUNE=50")
print("If Kaggle memory is low, keep USE_EFFICIENTNET=False (default) and/or USE_XCEPTION=False.")
```
